```python
import pandas as pd
```

```python
def _import_dependencies():
        from mlxtend.frequent_patterns import apriori, association_rules
        from mlxtend.preprocessing import TransactionEncoder

        return apriori, association_rules, TransactionEncoder
```

```python
apriori, association_rules, TransactionEncoder = (
            _import_dependencies()
        )
```

```python
# Load dataset
data = pd.read_excel('data/samples/online_retail_II.xlsx')
data.head()
```

```python
transaction_col = 'Invoice'
item_col = 'Description'
min_support = 0.01
APRIORI_MAX_LEN = 4 
```

```python
data = data.dropna(subset=[transaction_col, item_col])
data[item_col] = data[item_col].astype(str).str.strip()
data = data[data[item_col] != ""]
```

```python
def _build_transactions(
        work: pd.DataFrame,
        transaction_col: str,
        item_col: str,
    ) -> list[list[str]]:
    grouped = work.groupby(transaction_col, sort=False)[item_col]
    return grouped.apply(lambda items: [str(i) for i in items.tolist()]).tolist()
```

```python
transactions = _build_transactions(data, transaction_col, item_col)
```

```python
transactions[:5]
```

```python
# Group products by invoice
transactions = data.groupby('Invoice')['Description'].apply(list).values.tolist()
# Preview the transactional data
transactions[:5]
```

```python
te = TransactionEncoder()
te_ary = te.fit(transactions).transform(transactions)
encoded = pd.DataFrame(te_ary, columns=te.columns_, dtype=bool)
```

```python
encoded.head()
```

```python
frequent_itemsets = apriori(
            encoded,
            min_support=min_support,
            use_colnames=True,
            max_len=APRIORI_MAX_LEN,
            low_memory=True,
        )
```

```python
frequent_itemsets.head()
```

```python
min_confidence = 0.5
MAX_RULES_STORED = 5_000 
```

```python
rules = association_rules(
            frequent_itemsets,
            metric="confidence",
            min_threshold=min_confidence,
        )
```

```python
rules
```

```python
rules = rules.sort_values(["lift", "confidence"], ascending=False).reset_index(drop=True)
```

```python
if len(rules) > MAX_RULES_STORED:
            rules = rules.head(MAX_RULES_STORED).reset_index(drop=True)
```

```python
def _itemset_label(value) -> str:
    if hasattr(value, "__iter__") and not isinstance(value, (str, bytes)):
        return ", ".join(sorted(str(v) for v in value))
    return str(value)
```

```python
def _format_itemsets(itemsets: pd.DataFrame) -> pd.DataFrame:
    out = itemsets.copy()
    out["items"] = out["itemsets"].map(_itemset_label)
    out["item_count"] = out["itemsets"].map(len)
    return out.drop(columns=["itemsets"])
```

```python
def _format_rules(rules: pd.DataFrame) -> pd.DataFrame:
    out = rules.copy()
    out["antecedents"] = out["antecedents"].map(_itemset_label)
    out["consequents"] = out["consequents"].map(_itemset_label)
    out["rule"] = out["antecedents"] + "  →  " + out["consequents"]
    ordered = ["rule", "antecedents", "consequents"]
    for col in [
            "antecedent support", "consequent support",
            "support", "confidence", "lift", "leverage", "conviction",
        ]:
        if col in out.columns:
            ordered.append(col)
    return out[ordered]
```

```python
frequent_itemsets = frequent_itemsets.sort_values(
            ["support", "itemsets"], ascending=[False, True],
        ).reset_index(drop=True)

frequent_itemsets = _format_itemsets(frequent_itemsets)
rules = _format_rules(rules)
```
